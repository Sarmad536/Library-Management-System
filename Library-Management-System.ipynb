```python
class college:

    def __init__(self,name,location):
        self.name = name
        self.location = location

class student(college):

    def __init__(self,age,marks,name)
        self.age = age
        self.marks = marks
        super().__init__(name)

s1 = student(20,50,"SD")
print(s1.name,s1.marks, "/n", s1.age)
```

```python
class library:

    def __init__(self,budget):
        self.books = []
        self.books_no = 0
        self.budget = budget
        self.spend_amount = 0

    def add(self,book,price):
        if self.budget >= price:
            self.books.append(book)
            self.books_no += 1
            self.budget -= price
            self.spend_amount += price
            print(f"Congrats! You have bought {book}")
        else:
            print("you don't have enough money to buy this book")

    def remove(self,book,price):
        if book in self.books:
            self.books.remove(book)
            self.books_no -= 1
            self.budget += price
            self.spend_amount -= price
            print(f"you have returned {book}")
        else:
            print("you don't have that book")

    def show_books(self):
        print("Your library has these books:")
        for book in self.books:
            print(book)

    def count_books(self):
        print(f"you have total {self.books_no} books")

    def show_budget(self):
        print(f"you have left {self.budget} rupee")

    def show_spend_amount(self):
        print(f"you have spend {self.spend_amount} rupee")

budget = int(input("enter your budget: "))
l1 = library(budget)

while True:
    print("1.add books")
    print("2.remove books")
    print("3.show books")
    print("4.count books")
    print("5.show remaining budget")
    print("6.spend amount")
    print("7.exit")

    choice = int(input("enter your choice: "))

    if choice == 1:
        book = input("enter book name to add: ")
        price = int(input("enter price to buy book: "))
        l1.add(book,price)

    elif choice == 2:
        book = input("enter book name to remove: ")
        price = int(input("enter price to return book: "))
        l1.remove(book,price)

    elif choice == 3:
        l1.show_books()

    elif choice == 4:
        l1.count_books()

    elif choice == 5:
        l1.show_budget()

    elif choice == 6:
        l1.show_spend_amount()

    elif choice == 7:
        print("Thanks for visiting us")
        break
```
