```python
import random
no_games = int(input("enter how many rounds you want: "))
user_points = 0
comp_points = 0
lst = ["rock","paper","scissor"]
n = 1

while n <= no_games:
    print("                             ")

    print("----------new round----------")
    print(f"----------round {n}----------")
    print("                             ")

    user = input("enter your choice: ")
    comp = random.choice(lst)

    if user == "paper" and comp == "rock":
        print(f"you win this round bcz comp choice is {comp}")
        user_points += 1
        n += 1
        print("user points : comp points")
        print(user_points,"                ",comp_points)

    elif user == "paper" and comp == "scissor":
        print(f"you lose this round bcz comp choice is {comp}")
        comp_points += 1
        n += 1
        print("user points : comp points")
        print(user_points,"               ",comp_points)

    elif user == "rock" and comp == "paper":
        print(f"you lose this round bcz comp choice is {comp}")
        comp_points += 1
        n += 1
        print("user points : comp points")
        print(user_points,"                 ",comp_points)

    elif user == "rock" and comp == "scissor":
        print(f"you win this round bcz comp choice is {comp}")
        user_points += 1
        n += 1
        print("user points : comp points")
        print(user_points,"               ",comp_points)

    elif user == "scissor" and comp == "paper":
        print(f"you win this round bcz comp choice is {comp}")
        user_points += 1
        n += 1
        print("user points : comp points")
        print(user_points,"               ",comp_points)

    elif user == "scissor" and comp == "rock":
        print(f"you lose this round bcz comp choice is {comp}")
        comp_points += 1
        n += 1
        print("user points : comp points")
        print(user_points,"               ",comp_points)

    elif user == comp:
        print("draw bcz comp also choice same")
        n += 1
        print("user points : comp points")
        print(user_points,"               ",comp_points)

    else:
        print("Invalid input. Select from rock,paper or scissor")

print("                            ")
print("-----------result-----------")
print("                            ")

if user_points > comp_points:
    print("You win the battle.")
elif user_points < comp_points:
    print("You lose the battle")
else:
    print("It is draw")

print("user points : comp points")
print(user_points,"                 ",comp_points)

if user_points > comp_points:
    print("Now its time to continue the streak.Just enter play to strengthen the streak")
elif user_points < comp_points:
    print("Do you want revenge. Just enter play to put down this comp permanently")
else:
    print("Do you want to move one step more. Just enter play to become winner")
```
